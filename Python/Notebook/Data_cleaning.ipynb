# Data Cleaning

## Objective

Clean all datasets by handling missing values, removing duplicates, converting data types, validating data quality, and preparing the data for feature engineering.

```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
HOST = "localhost"
PORT = 3306
USER = "root"
PASSWORD = "root"
DATABASE = "retail_analysis"

engine = create_engine(
    f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}:{PORT}/{DATABASE}"
)
```

```python
customers = pd.read_sql("SELECT * FROM customers", engine)

orders = pd.read_sql("SELECT * FROM orders", engine)

products = pd.read_sql("SELECT * FROM products", engine)

order_items = pd.read_sql("SELECT * FROM order_items", engine)

payments = pd.read_sql("SELECT * FROM order_payments", engine)

reviews = pd.read_sql("SELECT * FROM order_reviews", engine)

sellers = pd.read_sql("SELECT * FROM sellers", engine)

geolocation = pd.read_sql("SELECT * FROM geolocation", engine)

category = pd.read_sql("SELECT * FROM category_translation", engine)
```

```python
tables = {
    "Customers": customers,
    "Orders": orders,
    "Products": products,
    "Order Items": order_items,
    "Payments": payments,
    "Reviews": reviews,
    "Sellers": sellers,
    "Geolocation": geolocation,
    "Category": category
}
```

```python
for name, df in tables.items():

    print("="*60)

    print(name)

    print("="*60)

    missing = df.isnull().sum()

    missing = missing[missing>0]

    print(missing)

    print()
```

# Duplicate Rows

```python
for name, df in tables.items():

    print(name)

    print("Duplicate Rows :",df.duplicated().sum())

    print()
```

```python
date_columns = [

'order_purchase_timestamp',

'order_approved_at',

'order_delivered_carrier_date',

'order_delivered_customer_date',

'order_estimated_delivery_date'

]

for col in date_columns:

    orders[col]=pd.to_datetime(
        orders[col]
    )
```

```python
reviews['review_creation_date']=pd.to_datetime(
reviews['review_creation_date']
)

reviews['review_answer_timestamp']=pd.to_datetime(
reviews['review_answer_timestamp']
)
```

```python
order_items['shipping_limit_date']=pd.to_datetime(
order_items['shipping_limit_date']
)
```

```python
# Remove Duplicate Rows

customers.drop_duplicates(inplace=True)

orders.drop_duplicates(inplace=True)

products.drop_duplicates(inplace=True)

order_items.drop_duplicates(inplace=True)

payments.drop_duplicates(inplace=True)

reviews.drop_duplicates(inplace=True)

sellers.drop_duplicates(inplace=True)

geolocation.drop_duplicates(inplace=True)

category.drop_duplicates(inplace=True)
```

```python
customers.to_csv(
'../outputs/cleaned_data/customers_clean.csv',
index=False
)

orders.to_csv(
'../outputs/cleaned_data/orders_clean.csv',
index=False
)

products.to_csv(
'../outputs/cleaned_data/products_clean.csv',
index=False
)

order_items.to_csv(
'../outputs/cleaned_data/order_items_clean.csv',
index=False
)

payments.to_csv(
'../outputs/cleaned_data/payments_clean.csv',
index=False
)

reviews.to_csv(
'../outputs/cleaned_data/reviews_clean.csv',
index=False
)

sellers.to_csv(
'../outputs/cleaned_data/sellers_clean.csv',
index=False
)

geolocation.to_csv(
'../outputs/cleaned_data/geolocation_clean.csv',
index=False
)

category.to_csv(
'../outputs/cleaned_data/category_translation_clean.csv',
index=False
)
```
